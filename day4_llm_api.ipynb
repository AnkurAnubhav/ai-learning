```python
from dotenv import load_dotenv
import os

load_dotenv()

api_key = os.getenv("ANTHROPIC_API_KEY")

if api_key:
    print("API key loaded successfully!")
    print(f"Key starts with: {api_key[:10]}...")
else:
    print("API key not found — check your .env file")
```

```python
import anthropic

client = anthropic.Anthropic()

message = client.messages.create(
    model = "claude-haiku-4-5-20251001",
    max_tokens=1024,
    messages=[
        {"role": "user", "content": "Which is the longest spanning empire in history?"}
    ]
)

print(message.content[0].text)
```

```python
import anthropic

client = anthropic.Anthropic()

message = client.messages.create(
    model="claude-haiku-4-5-20251001",
    max_tokens=1024,
    system="You are a profession financial analyst.",
    messages=[
        {"role": "user", 
         "content": "What do you think about the fundamentals of the company SpaceX? Is it a good buy, sell or hold"}
    ]
)
print(message.content[0].text)
```

```python
import anthropic

client = anthropic.Anthropic()

messages = []

question = input("first question: ")
while(question != "quit"):
    newMessage = {"role": "user", "content": question}
    messages.append(newMessage)

    message = client.messages.create(
        model="claude-haiku-4-5-20251001",
        max_tokens=1024,
        messages = messages
    )

    response = message.content[0].text
    print(f"Claude: {response}")

    messages.append({"role": "assistant", "content": response})

    question = input(question)

print(message.content[0].text)
```
